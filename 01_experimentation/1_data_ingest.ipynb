```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

# [1] Load Data

```python
data_original = pd.read_csv("exp_data/original_data.csv")
```

# [2] Split into Spans
This is just an example of splitting data into spans. In real production scenarios, data spans are not obtained this way. Rather they're obtained based on new incoming data.

```python
span1, span2 = train_test_split(data_original, test_size=0.5, random_state=42)
```

# [3] Split Span 1
Assuming we only have Span 1; overwhich we will workout our analysis.

```python
train1, temp1 = train_test_split(span1, test_size=0.30, random_state=42)  # 70% train, 30% temp
test1, eval1 = train_test_split(temp1, test_size=0.50, random_state=42)   # 15% test, 15% eval
```

# [4] Split Span 2
Assuming Span 2 will come later in the future. Will not be used now.

```python
train2, temp2 = train_test_split(span2, test_size=0.30, random_state=42)
test2, eval2 = train_test_split(temp2, test_size=0.50, random_state=42)
```

# [5] Export Results

```python
# Span Sizes
print("Span1 sizes:", len(train1), len(test1), len(eval1))
print("Span2 sizes:", len(train2), len(test2), len(eval2))
```

```python
# Span 1 and Span 2
span1.to_csv("exp_data/span1.csv", index=False)
span2.to_csv("exp_data/span2.csv", index=False)
```

```python
from pathlib import Path

# Define Base Paths
base_path = Path("exp_data/span-1")

(train_path := base_path / "train").mkdir(parents=True, exist_ok=True)
(test_path  := base_path / "test").mkdir(parents=True, exist_ok=True)
(eval_path  := base_path / "eval").mkdir(parents=True, exist_ok=True)

# Span 1 Train-Test-Eval Data
train1.to_csv("exp_data/span-1/train/train.csv", index=False)
test1.to_csv("exp_data/span-1/test/test.csv", index=False)
eval1.to_csv("exp_data/span-1/eval/eval.csv", index=False)
```

```python
# Define Base Paths
base_path2 = Path("exp_data/span-2")

(train_path := base_path2 / "train").mkdir(parents=True, exist_ok=True)
(test_path  := base_path2 / "test").mkdir(parents=True, exist_ok=True)
(eval_path  := base_path2 / "eval").mkdir(parents=True, exist_ok=True)

# Span 2 Train-Test-Eval Data
train2.to_csv("exp_data/span-2/train/train.csv", index=False)
test2.to_csv("exp_data/span-2/test/test.csv", index=False)
eval2.to_csv("exp_data/span-2/eval/eval.csv", index=False)
```
